## Pandas Series
A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.<br>
there are multiple ways to create a pandas series :

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a', 'b', 'c'] # list 
lists = [10, 20, 30] # list 
arr = np.array([10, 20, 30]) # array
d = {1 : 10, 2 : 20, 3 : 30}  # distonary
```

```python
list_series = pd.Series(lists) # series of lists
list_series # output -> veritcally first index & second data points
```

```python
custom_series = pd.Series(lists, index = labels) # custom index as labels data
custom_series
```

```python
arr_series = pd.Series(arr) # series of arr
arr_series
```

```python
dist_series = pd.Series(d)  # series of dictionary(d)
dist_series  # auto dectect the index of dictionary
```
