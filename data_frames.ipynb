```python
import numpy as np
import pandas as pd
```

- ***creating a data frame***

```python
data_dist = { # data dictionary
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}

df_dist = pd.DataFrame(data_dist)
df_dist # output -> table format with headers name & 0 based indexing
```
