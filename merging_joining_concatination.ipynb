```python
import numpy as np
import pandas as pd
```

- ***merging 2 dataframes***

```python
employees = pd.DataFrame(
    {
        "employee_id" : [1, 2, 3, 4, 5],
        "name" : ["John", "Anna", "Peter", "Linda", "Bob"],
        "department" : ["HR", "IT", "Finance", "IT", "HR"],
    }
)

salaries = pd.DataFrame(
    {
        "employee_id" : [1, 2, 3, 6, 7],
        "salary" : [60000, 80000, 65000, 70000, 90000],
        "bonus" : [5000, 10000, 7000, 8000, 12000],
    }
)
```

```python
employees
```

```python
salaries
```

```python
merged_inner_df = pd.merge(employees, salaries) # common values
merged_inner_df
```

```python
merged_inner_df = pd.merge(employees, salaries, how = 'outer')  # all values
merged_inner_df
```

```python
merged_left_df = pd.merge(employees, salaries, how = 'left')  # merge on left basis
merged_left_df
```

```python
merged_right_df = pd.merge(employees, salaries, how = 'right')  # merge on right basis
merged_right_df
```

- ***concatination of 2 dataframes***

```python
df1 = pd.DataFrame(
    {
        "A": ["A0", "A1", "A2"], 
        "B": ["B0", "B1", "B2"], 
        "C": ["C0", "C1", "C2"]
    }
)

df2 = pd.DataFrame(
    {
        "A": ["A3", "A4", "A5"], 
        "B": ["B3", "B4", "B5"], 
        "C": ["C3", "C4", "C5"]
    }
)
```

```python
df1
```

```python
df2
```

```python
concat_col = pd.concat([df2, df1]) # combine df in column
concat_col 
```

```python
concat_row = pd.concat([df2, df1], axis = 1)  # combine df in row
concat_row
```

- ***joining of 2 dataframes***

```python
join_df1 = pd.DataFrame(
    {
        "name": ["Alice", "Bob", "Charlie"]
    }, 
    index=[1, 2, 3]
)

join_df2 = pd.DataFrame(
    {
        "score": [85, 90, 75]
    },
    index=[2, 3, 4]
)
```

```python
join_df1
```

```python
join_df2
```

```python
join_df1.join(join_df2) # join the df based on common index
```

```python
join_df1.join(join_df2, how = 'outer')  # join the df based on all index
```
